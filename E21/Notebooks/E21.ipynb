#### Clase Producto: 
Define un producto cualquiera con su nombre y precio, siendo este último un atributo protegido al cual le definimos un getter con @property y un setter con @nombreatributo.setter.

##### @property precio:
Este método (¿?) es el getter, desde mi punto de vista actual @property lo que hace es convertir un atributo en un atributo con getter y setter para controlar las modificaciones y los accesos a la información de tal atributo, en este código se utiliza para simplemente devolver el precio del producto.

##### @precio.setter precio:
Análogamente a @property precio, no sé si concluir si es un método, lo que sí sé es que al haber construido un getter se tiene que definir un setter para controlar las modificaciones del atributo en cuestión, en este caso se verifica que sea mayor o igual a 0 el precio nuevo del producto y luego se modifica directamente el atributo ya que gracias a @precio.setter podemos acceder directamente al atributo protegido como un atributo sin protección y se imprime en pantalla el precio al que se cambió.


```python

class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self._precio = precio

    @property
    def precio(self):
        return self._precio

    @precio.setter
    def precio(self, precio_nuevo):
        if precio_nuevo <= 0:
            print("Ingresa un precio mayor a 0. Intentalo de nuevo")
            return
        self._precio = precio_nuevo
        print(f"Se ha cambiado exitosamente el precio a {self._precio}")
        return
```

### Flujo del programa

Se define un producto pera, se busca su precio para verificar el funcionamiento del getter, y luego se hacen tres intentos de modificación los cuales dos resultan fallidos al no cumplir las claúsulas de guarda del getter, lo que protege al atributo protegido de introducir datos erróneos para luego hacer una modificación exitosa a un precio de 1000.

```python

pera = Producto("pera", 100)

print(pera.precio)

pera.precio = -10
pera.precio = 0
pera.precio = 1000
```
