CLASES

Clase Persona: Esta clase permite definir personas genéricas, además permite crear clases hijas de personas con roles específicos y instancias de personas específicas

Método obtener_rol: Imprime el rol de la instancia, en clases hijas se utiliza polimorfismo para que muestre el rol de esas clases

```python
class Persona:
    def __init__(self, nombre, cedula):
        self.nombre = nombre
        self.cedula = cedula

    def obtener_rol(self):
        print("Persona genérica")
        return
```

Clase Administrativo: Es una clase hija de Persona, sirve para crear instancias que tengan el rol Administrativo, además esas instancias tienen una dependencia asignada

Método obtener_rol: Imprime el rol de la instancia y su dependencia, el método fue sobreescrito de la clase Persona

```python
class Administrativo(Persona):
    def __init__(self, nombre, cedula, dependencia):
        super().__init__(nombre, cedula)
        self.dependencia = dependencia
        
    def obtener_rol(self):
        print(f"Administrativo en {self.dependencia}")
        return
```

Clase Docente: Es una clase hija de Persona, sirva para crear instancias que tengan el rol Docente, además esas instancias tienen una materia asignada

Método obtener_rol: Imprime el rol de la instancia y su materia_dictada, el método fue sobre escrito de la clase Persona 

```python
class Docente(Persona):
    def __init__(self, nombre, cedula, materia_dictada):
        super().__init__(nombre, cedula)
        self.materia_dictada = materia_dictada

    def obtener_rol(self):
        print(f"Docente de {self.materia_dictada}")
        return
```
