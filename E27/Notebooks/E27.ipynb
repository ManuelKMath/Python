```python
"""Clases de excepciones.py"""

class FinanzasError(Exception):
    """Clase base de errores"""
    pass

class ActivoExistenteError(FinanzasError):
    """Clase que se encarga del error ActivoExistenteError, este error se lanza cuando ya hay un activo igual en la lista de activos"""
    def __init__(self, simbolo: str):
        self.simbolo = simbolo
        super().__init__(f"[ERROR]: ACTIVO YA EXISTENTE")

class ActivoNoEncontrado(FinanzasError):
    """Clase que se encarga del error ActivoNoEncontrado, este error se lanza cuando no hay un activo igual en la lista de activos"""
    def __init__(self, simbolo: str):
        self.simbolo = simbolo
        super().__init__(f"[ERROR]: ACTIVO NO ENCONTRADO")

class FondosInsuficientesError(FinanzasError):
    """Clase que se encarga del error FondosInsuficientesError, este error se lanza cuando no hay suficiente saldo para concretar una transacción"""
    def __init__(self, monto_requerido : float, saldo_disponible: float):
        self.monto_requerido = monto_requerido
        self.saldo_disponible = saldo_disponible
        super().__init__(f"[ERROR]: FONDOS INSUFICIENTES")

class OperacionInvalidaError(FinanzasError):
    def __init__(self):
        super().__init__("[ERROR]: OPERACIÓN INVALIDA")

"""Clases de estrategias.py"""

from abc import ABC, abstractmethod

class EstrategiaComision(ABC):
    """Clase abstracta que permite implementar dos tipos de comisiones distintas, una fija y una porcentual"""
    @abstractmethod
    def calcular_comision(self, monto: float) -> float:
        """Método abstracto que obliga a usarse en las implementaciones: ComisionFija y ComisionPorcentual"""
        pass

class ComisionFija(EstrategiaComision):
    def calcular_comision(self, monto: float) -> float:
        """Método heredado de EstrategiaComision y sobreescrito para devolver un valor fijo"""
        return 5.0

class ComisionPorcentual(EstrategiaComision):
    def calcular_comision(self, monto: float) -> float:
        """Método heredado de EstrategiaComision y sobreescrito para devolver un valor porcentual"""
        comision = monto * 0.015
        if comision <= 2.0:
            return 2.0
        return comision

"""Clases de modelos.py"""

from abc import ABC, abstractmethod
from estrategias import EstrategiaComision
from excepciones import OperacionInvalidaError

class ActivoFinanciero(ABC):
    """Clase abstracta que permite implementar tres tipos de activos financieros: Accion, Criptomoneda y Bono"""
    def __init__(self, simbolo: str, nombre: str, precio_actual: float, cantidad: float, estrategia_comision: EstrategiaComision):
        self._simbolo = simbolo
        self._nombre = nombre
        self._precio_actual = precio_actual
        self._cantidad = cantidad
        self._estrategia_comision = estrategia_comision

    @property
    def simbolo(self) -> str:
        """Getter del atributo protegido simbolo"""
        return self._simbolo

    @property
    def precio_actual(self) -> float:
        """Getter del atributo protegido precio_actual"""
        return self._precio_actual

    @precio_actual.setter
    def precio_actual(self, nuevo_precio: float) -> None:
        """Setter del atributo protegido precio_actual: Valida si al modificar el precio este es positivo."""
        if nuevo_precio <= 0:
            raise OperacionInvalidaError
        self._precio_actual = nuevo_precio
        return
        
    @property
    def cantidad(self) -> float:
        """Getter del atributo protegido cantidad"""
        return self._cantidad

    @cantidad.setter
    def cantidad(self, nueva_cantidad: float) -> None:
        """Setter del atributo protegido cantidad: Valida si al modificar la cantidad ésta es positiva."""
        if nueva_cantidad <= 0:
            raise OperacionInvalidaError
        self._cantidad = nueva_cantidad
        return

    @abstractmethod
    def calcular_valor_total(self) -> float:
        """Método abstracto que obliga a usarse en las implementaciones: Accion, Criptomoneda y Bono"""
        pass

    @abstractmethod
    def evaluar_riesgo(self) -> str:
        """Método abstracto que obliga a usarse en las implementaciones: Accion, Criptomoneda y Bono"""
        pass

    @abstractmethod
    def calcular_valor(self) -> float:
        """Método abstracto que obliga a usarse en las implementaciones: Accion, Criptomoneda y Bono"""
        pass

    def __repr__(self) -> str:
        """Método nativo que da información técnica del objeto"""
        return f"Activo(Simbolo = \'{self._simbolo}\', Cantidad = {self._cantidad})"

    def __eq__(self, otro: object) -> bool:
        """Método nativo que permite definir igualdades entre objetos, distintas a la nativa que compara la id en memoria"""
        if not isinstance(otro, ActivoFinanciero):
            raise OperacionInvalidaError
        simbolo_1 = self._simbolo
        simbolo_2 = otro.simbolo
        if simbolo_1.lower() != simbolo_2. lower():
            pass
        return True

    def __str__(self) -> str:
        return "Activo"

class Accion(ActivoFinanciero):
    "Clase de los objetos acción, calcula el valor total de ellos y retorna su riesgo"
    def __init__(self, simbolo: str, nombre: str, precio_actual: float, cantidad: float, estrategia_comision: EstrategiaComision, sector: str):
        super().__init__(simbolo, nombre, precio_actual, cantidad, estrategia_comision)
        self._sector = sector

    def calcular_valor_total(self) -> float:
        return (self._cantidad*self._precio_actual)+self._estrategia_comision.calcular_comision(self._cantidad*self._precio_actual)
    
    def evaluar_riesgo(self) -> str:
        return "Riesgo Moderado"
    
    def calcular_valor(self, cantidad: float) -> float:
        return (cantidad*self._precio_actual)+self._estrategia_comision.calcular_comision(cantidad*self._precio_actual)

    def __str__(self):
        return f"ID: {self._simbolo} | Empresa: {self._nombre} | Precio Unitario: {self._precio_actual} | Cantidad: {self._cantidad} | Sector: {self._sector}"

class Criptomoneda(ActivoFinanciero):
    "Clase de los objetos criptomoneda, calcula el valor total de ellos y retorna su riesgo"
    def __init__(self, simbolo: str, nombre: str, precio_actual: float, cantidad: float, estrategia_comision: EstrategiaComision, volatilidad: float):
        super().__init__(simbolo, nombre, precio_actual, cantidad, estrategia_comision)
        self._volatilidad = volatilidad

    def calcular_valor_total(self) -> float:
        return (self._cantidad*self._precio_actual)+self._estrategia_comision.calcular_comision(self._cantidad*self._precio_actual)
    
    def evaluar_riesgo(self) -> str:
        if self._volatilidad > 0.5:
            return "Riesgo Alto"
        return "Riesgo Moderado"

    def calcular_valor(self, cantidad: float) -> float:
        return (cantidad*self._precio_actual)+self._estrategia_comision.calcular_comision(cantidad*self._precio_actual)

    def __str__(self):
        return f"ID: {self._simbolo} | Nombre: {self._nombre} | Precio Unitario: {self._precio_actual} | Cantidad: {self._cantidad} | Volatilidad: {self._volatilidad}"

class Bono(ActivoFinanciero):
    "Clase de los objetos Bono, calcula el valor total de ellos y retorna su riesgo"
    def __init__(self, simbolo: str, nombre: str, precio_actual: float, cantidad: float, estrategia_comision: EstrategiaComision, tasa_interes: float):
        super().__init__(simbolo, nombre, precio_actual, cantidad, estrategia_comision)
        self._tasa_interes = tasa_interes

    def calcular_valor_total(self) -> float:
        return self._cantidad*self._precio_actual*(1+self._tasa_interes)+self._estrategia_comision.calcular_comision(self._cantidad*self._precio_actual*(1+self._tasa_interes))
    
    def evaluar_riesgo(self) -> str:
        return "Riesgo Bajo"

    def calcular_valor(self, cantidad: float) -> float:
        return cantidad*self._precio_actual*(1+self._tasa_interes)+self._estrategia_comision.calcular_comision(cantidad*self._precio_actual*(1+self._tasa_interes))
    
    def __str__(self):
        return f"ID: {self._simbolo} | Empresa: {self._nombre} | Precio Unitario: {self._precio_actual} | Cantidad: {self._cantidad} | Tasa de interés: {self._tasa_interes}"

"""Clases de portafolio.py"""

from excepciones import ActivoExistenteError, ActivoNoEncontrado, FondosInsuficientesError, OperacionInvalidaError
from modelos import ActivoFinanciero

class Portafolio:
    "Clase del objeto Portafolio, donde se almacenan los objetos de las subclases de ActivoFinanciero en una lista, y la que realiza las operaciones con ellos"
    def __init__(self, activos: list, saldo_efectivo: float):
        self._activos = activos
        self._saldo_efectivo = saldo_efectivo

    @property
    def saldo_efectivo(self) -> float:
        """Getter del atributo protegido saldo_efectivo"""
        return self._saldo_efectivo
    
    @saldo_efectivo.setter
    def saldo_efectivo(self, saldo_nuevo: float) -> None:
        """Setter del atributo protegido saldo_efectivo"""
        if saldo_nuevo <= 0:
            raise OperacionInvalidaError
        self._saldo_efectivo += saldo_nuevo
        return
        
    def agregar_activo(self, activo: ActivoFinanciero) -> None:
        """Método que permite agregar activos a la estructura de datos dentro de la instancia Portafolio"""
        if not isinstance(activo, ActivoFinanciero):
            raise OperacionInvalidaError
        if activo in self._activos:
            raise ActivoExistenteError(activo.simbolo)
        if self._saldo_efectivo < activo.calcular_valor_total():
            raise FondosInsuficientesError(activo.calcular_valor_total(), self._saldo_efectivo)
        self._saldo_efectivo -= activo.calcular_valor_total()
        self._activos.append(activo)
        return

    def comprar_activo(self, simbolo: str, cantidad_adicional: float) -> None:
        """Método que permite comprar más activos que ya están dentro de la instancia Portafolio"""
        if cantidad_adicional <= 0:
            raise OperacionInvalidaError
        for activo in self._activos:
            if activo.simbolo == simbolo:
                if self._saldo_efectivo < activo.calcular_valor(cantidad_adicional):
                    raise FondosInsuficientesError(activo.calcular_valor(cantidad_adicional), self._saldo_efectivo)
                activo.cantidad += cantidad_adicional
                self._saldo_efectivo -= activo.calcular_valor(cantidad_adicional)
                return
        raise ActivoNoEncontrado(simbolo)

    def calcular_valor_total_portafolio(self) -> float:
        """Método que devuelve todo el valor del Portafolio"""
        valor_total_portafolio = 0
        for activo in self._activos:
            valor_total_portafolio += activo.calcular_valor_total()
        valor_total_portafolio += self._saldo_efectivo
        return valor_total_portafolio

    def __len__(self) -> int:
        """Método nativo que permite usar la función len() en la instancia Portafolio"""
        return len(self._activos)

    def __getitem__(self, posicion: int) -> ActivoFinanciero:
        """Método nativo que permite indexar y convertir en iterable la instancia Portafolio"""
        return self._activos[posicion]

"""Main.py"""

if __name__ == "__main__":
    import pickle
    from pathlib import Path
    from estrategias import ComisionFija, ComisionPorcentual
    from excepciones import ActivoExistenteError, ActivoNoEncontrado, FondosInsuficientesError, OperacionInvalidaError
    from modelos import Accion, Criptomoneda, Bono
    from portafolio import Portafolio
    
    def guardar_datos(ruta, datos: Portafolio) -> None:
        """Función de guardado de datos"""
        with open(ruta, "wb") as archivo:
            pickle.dump(datos, archivo)
        return

    def cargar_datos(ruta) -> Portafolio:
        """Función de carga de datos"""
        try:
            with open(ruta, "rb") as archivo: 
                return pickle.load(archivo)
        except (FileNotFoundError, pickle.PickleError):
            return Portafolio([], 0)    

    def menu() -> None:
        "Función que se encarga del menú principal del programa"
        while True:
            print("\n---Menú---")
            print("1. Agregar activo al portafolio")
            print("2. Comprar más activos")
            print("3. Calcular valor total del portafolio")
            print("4. Guardar cambios")
            print("5. Añadir saldo al portafolio")
            print("6. Ver activos en el portafolio")
            print("7. Salir")
            try:
                opcion = float(input("Introduce una de las opciones mostradas en el menú: "))
            except ValueError:
                print("Introduce uno de los números que aparecen para elegir.")
                continue
            match opcion:
                case 1:
                    """Agregar activo"""
                    menu_1()
                    continue
                case 2:
                    """Comprar activo"""
                    menu_2()
                    continue
                case 3:
                    """Calcular valor total del portafolio"""
                    print(f"Valor total del portafolio: {portafolio_base.calcular_valor_total_portafolio()}$")
                    continue
                case 4: 
                    guardar_datos(ruta, portafolio_base)
                    print("Datos guardados exitosamente")
                    continue
                case 5:
                    try:
                        saldo_nuevo = float(input("Introduce el saldo que vas a añadir al portafolio: "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.saldo_efectivo = saldo_nuevo
                    except OperacionInvalidaError:
                        print("Introduce un número mayor a 0.")
                        continue
                    print("Saldo añadido exitosamente")
                    continue
                case 6:
                    print("---Porfafolio---")
                    for activo in portafolio_base:
                        print(activo)
                    continue
                case 7:
                    """Salir del programa"""
                    print("Saliendo...")
                    break
                case _:
                    """Cualquier otro caso"""
                    print("Introduce uno de los números que aparecen para elegir.")
                    continue

    def menu_1():
        """Menú que se encarga de gestionar la adición por primera vez de activos al portafolio"""
        while True:
            print("\n1. Agregar acciones al portafolio")
            print("2. Agregar criptomonedas al portafolio")
            print("3. Agregar bonos al portafolio")
            print("4. Volver al menú anterior")
            try:
                opcion_1 = float(input("Introduce una de las opciones mostradas en el menú: "))
            except ValueError:
                print("Introduce uno de los números que aparecen para elegir.")
                continue
            match opcion_1:
                case 1:
                    """Agregar acciones"""
                    simbolo = input("¿Cuál es el ID de las acciones? ")
                    nombre = input("¿Cuál es el nombre de la empresa? ")
                    try:
                        precio_actual = float(input("¿Cuál es el precio unitario de las acciones de esta empresa? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    respuestas_comision_fija = ["fija", "comisión fija", "comision fija"]
                    respuestas_comision_porcentual = ["porcentual", "comisión porcentual", "comision porcentual"]
                    while True:
                        comision_opcion = input("¿La comisión es fija o porcentual? ")
                        if comision_opcion.lower() in respuestas_comision_fija:
                            estrategia_comision = ComisionFija()
                            break
                        if comision_opcion.lower() in respuestas_comision_porcentual:
                            estrategia_comision = ComisionPorcentual()
                            break
                        print("Intente de nuevo. Si es comisión fija, responda fija, si es comisión porcentual responda porcentual.")
                        continue
                    sector = input("¿A cuál sector pertenece la empresa de estas acciones? ")
                    try:
                        portafolio_base.agregar_activo(Accion(simbolo, nombre, precio_actual, cantidad, estrategia_comision, sector))
                    except ActivoExistenteError as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregadas {cantidad} acciones de la empresa {nombre} con ID: {simbolo} al portafolio")
                    continue
                case 2:
                    """Agregar criptomonedas"""
                    simbolo = input("¿Cuál es el ID de la esta criptomoneda?")
                    nombre = input("¿Cuál es el nombre de la criptomoneda?")
                    try:
                        precio_actual = float(input("¿Cuál es el precio unitario de la criptomoneda? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    respuestas_comision_fija = ["fija", "comisión fija", "comision fija"]
                    respuestas_comision_porcentual = ["porcentual", "comisión porcentual", "comision porcentual"]
                    while True:
                        comision_opcion = input("¿La comisión es fija o porcentual? ")
                        if comision_opcion.lower() in respuestas_comision_fija:
                            estrategia_comision = ComisionFija()
                            break
                        if comision_opcion.lower() in respuestas_comision_porcentual:
                            estrategia_comision = ComisionPorcentual()
                            break
                        print("Intente de nuevo. Si es comisión fija, responda fija, si es comisión porcentual responda porcentual.")
                        continue
                    try:
                        volatilidad = float(input("¿Cuál es la volatilidad de esta criptomoneda? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.agregar_activo(Criptomoneda(simbolo, nombre, precio_actual, cantidad, estrategia_comision, volatilidad))
                    except ActivoExistenteError as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregadas {cantidad} criptomonedas de nombre {nombre} con ID: {simbolo} al portafolio")
                    continue
                case 3:
                    """Agregar Bono"""
                    nombre = input("¿Cuál es el nombre de la empresa? ")
                    try:
                        precio_actual = float(input("¿Cuál es el precio unitario de los bonos de esta empresa? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    respuestas_comision_fija = ["fija", "comisión fija", "comision fija"]
                    respuestas_comision_porcentual = ["porcentual", "comisión porcentual", "comision porcentual"]
                    while True:
                        comision_opcion = input("¿La comisión es fija o porcentual? ")
                        if comision_opcion.lower() in respuestas_comision_fija:
                            estrategia_comision = ComisionFija()
                            break
                        if comision_opcion.lower() in respuestas_comision_porcentual:
                            estrategia_comision = ComisionPorcentual()
                            break
                        print("Intente de nuevo. Si es comisión fija, responda fija, si es comisión porcentual responda porcentual.")
                        continue
                    try:
                        tasa_interes = float(input("¿Cuál es la tasa de interés de los bonos? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.agregar_activo(Bono(simbolo, nombre, precio_actual, cantidad, estrategia_comision, tasa_interes))
                    except ActivoExistenteError as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregados {cantidad} bonos de la empresa {nombre} con ID: {simbolo} al portafolio")
                    continue
                case 4:
                    """Salir del submenú"""
                    print("Volviendo al menú anterior...")
                    break
                case _:
                    """Cualquier otro caso"""
                    print("Introduce uno de los números que aparecen para elegir.")
                    continue

    def menu_2():
        """Menú que se encarga de gestionar las compras de activos al portafolio"""
        while True:
            print("\n1. Comprar acciones")
            print("2. Comprar criptomonedas")
            print("3. Comprar bonos")
            print("4. Volver al menú anterior")
            try:
                opcion_2 = float(input("Introduce una de las opciones mostradas en el menú: "))
                if not isinstance(opcion_2, float):
                    raise TypeError
            except TypeError:
                print("Introduce uno de los números que aparecen para elegir.")
                continue
            match opcion_2:
                case 1:
                    """Comprar acciones"""
                    simbolo = input("¿Cuál es el ID de las acciones? ")
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.comprar_activo(simbolo, cantidad)
                    except ActivoNoEncontrado as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregadas {cantidad} acciones con ID: {simbolo} al portafolio")
                    continue
                case 2:
                    """Comprar criptomonedas"""
                    simbolo = input("¿Cuál es el ID de las criptomonedas? ")
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.comprar_activo(simbolo, cantidad)
                    except ActivoNoEncontrado as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregadas {cantidad} con ID: {simbolo} al portafolio")
                    continue
                case 3:
                    """Comprar bonos"""
                    simbolo = input("¿Cuál es el ID de los bonos? ")
                    try:
                        cantidad = float(input("¿Cuántas desea comprar? "))
                    except ValueError:
                        print("Introduce un número mayor a 0.")
                        continue
                    try:
                        portafolio_base.agregar_activo(simbolo, cantidad)
                    except ActivoExistenteError as e:
                        print(e)
                        print(f"ID del activo: {e.simbolo}")
                        continue
                    except FondosInsuficientesError as e:
                        print(e)
                        print(f"Saldo actual: {e.saldo_disponible} | Monto requerido: {e.monto_requerido}")
                        continue
                    except OperacionInvalidaError as e:
                        print(e)
                        continue
                    print(f"Han sido agregados {cantidad} con ID: {simbolo} al portafolio")
                    continue
                case 4:
                    """Salir del submenú"""
                    print("Volviendo al menú anterior...")
                    break
                case _:
                    """Cualquier otro caso"""
                    print("Introduce uno de los números que aparecen para elegir.")
                    continue

"""Flujo del programa"""

"""Cada vez que inicio el programa obtengo la ruta del archivo con Path, y busco la carpeta `Datos` y el archivo `datos_portafolio.pkl`
que es ahí donde se va a cargar y guardar la instancia Portafolio"""
ruta_archivo = Path(__file__).resolve()
ruta_raiz = ruta_archivo.parent
ruta = ruta_raiz / "Datos" / "datos_portafolio.pkl"
portafolio_base = cargar_datos(ruta)
"""Luego, se abre el menú donde el usuario interactúa con su portafolio."""
menu()
```
